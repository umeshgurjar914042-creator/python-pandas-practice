### 1. Integer (पूर्णांक संख्या)
जब हमें बिना दशमलव (decimal) वाले पूरे नंबर स्टोर करने होते हैं।

```python
# उदाहरण 1: उम्र (Integer)
umar = 25
print("उम्र:", umar)
print("Data Type:", type(umar))
```

### 2. Float (दशमलव संख्या)
जब हमें दशमलव वाले नंबर जैसे कि कीमत या तापमान स्टोर करना हो।

```python
# उदाहरण 2: तापमान (Float)
tapman = 36.6
print("तापमान:", tapman)
print("Data Type:", type(tapman))
```

### 3. String (टेक्स्ट/अक्षर)
जब हमें कोई नाम, वाक्य या शब्द स्टोर करना हो। इसे हमेशा सिंगल (' ') या डबल (" ") कोट्स के अंदर लिखा जाता है।

```python
# उदाहरण 3: नाम (String)
naam = "राम"
print("नाम:", naam)
print("Data Type:", type(naam))
```

### 4. Boolean (सत्य या असत्य)
यह केवल दो ही मान (values) ले सकता है: `True` (हाँ/सही) या `False` (ना/गलत)।

```python
# उदाहरण 4: क्या यूज़र लॉग इन है? (Boolean)
is_logged_in = True
print("लॉग इन स्थिति:", is_logged_in)
print("Data Type:", type(is_logged_in))
```

### 5. List (चीज़ों की सूची)
जब हमें एक ही variable में एक से ज़्यादा चीज़ें एक क्रम में रखनी हों। इसे `[]` ब्रैकेट्स में लिखा जाता है।

```python
import pandas as pd

# डेटा तैयार करना
data = {
    "फल": ["आम", "केला", "सेब", "अंगूर"],
    "कीमत": [80, 40, 120, 90],
    "मात्रा": [10, 25, 5, 8]
}

# DataFrame बनाना
df = pd.DataFrame(data)

# 'कुल' कॉलम बनाना (कीमत * मात्रा)
df["कुल"] = df["कीमत"] * df["मात्रा"]

# DataFrame प्रिंट करना
print("--- फलों का डेटाफ़्रेम ---")
print(df)
print()

# कुल बिक्री का जोड़
total_sales = df["कुल"].sum()
print("कुल बिक्री का जोड़ (Total Sales):", total_sales)
```

```python
# 50 से ज़्यादा कीमत वाले फलों को फ़िल्टर करना
mehnge_fal = df[df["कीमत"] > 50]

print("--- 50 से ज़्यादा कीमत वाले फल ---")
display(mehnge_fal)
```

```python
import pandas as pd
import random

# नकली डेटा (Synthetic Data) के लिए विकल्प
products = ["लैपटॉप", "मोबाइल", "हेडफोन", "स्मार्टवॉच", "टैबलेट"]
cities = ["दिल्ली", "मुंबई", "बेंगलुरु", "कोलकाता", "चेन्नई"]

# 10 पंक्तियों का डेटा तैयार करना
random.seed(42)  # परिणाम को सुसंगत रखने के लिए
sales_data = {
    "प्रोडक्ट": [random.choice(products) for _ in range(10)],
    "शहर": [random.choice(cities) for _ in range(10)],
    "कीमत": [random.randint(1000, 50000) for _ in range(10)],
    "मात्रा": [random.randint(1, 5) for _ in range(10)]
}

# DataFrame बनाना
sales_df = pd.DataFrame(sales_data)

# sales.csv नाम से सेव करना
sales_df.to_csv("sales.csv", index=False)

print("--- sales.csv सफलतापूर्वक बना दिया गया है ---")
display(sales_df)
```

```python
# sales.csv फ़ाइल को लोड करना
loaded_df = pd.read_csv("sales.csv")

# पहली 5 पंक्तियाँ दिखाना
print("--- sales.csv की पहली 5 पंक्तियाँ ---")
display(loaded_df.head(5))
```

```python
# DataFrame की संरचना और डेटा टाइप्स की जानकारी (info)
print("--- DataFrame की मूलभूत जानकारी (Info) ---")
loaded_df.info()

print("\n" + "="*50 + "\n")

# DataFrame का सांख्यिकीय विवरण (describe)
print("--- DataFrame का सांख्यिकीय विवरण (Describe) ---")
display(loaded_df.describe())
```

```python
# हर रिकॉर्ड के लिए कुल बिक्री (कीमत * मात्रा) की गणना करना
loaded_df["कुल बिक्री"] = loaded_df["कीमत"] * loaded_df["मात्रा"]

# शहर के अनुसार कुल बिक्री का योग निकालना
city_sales = loaded_df.groupby("शहर")["कुल बिक्री"].sum().reset_index()

# कुल बिक्री के आधार पर अवरोही क्रम (descending order) में सॉर्ट करना
city_sales = city_sales.sort_values(by="कुल बिक्री", ascending=False)

print("--- शहर के अनुसार कुल बिक्री ---")
display(city_sales)

# सबसे ज़्यादा बिक्री वाला शहर
top_city = city_sales.iloc[0]
print(f"\nसबसे ज़्यादा बिक्री वाला शहर: {top_city['शहर']} ({top_city['कुल बिक्री']} रुपए)")
```
